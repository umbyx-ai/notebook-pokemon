## Pregunta 1
##### De cada hábitat, cual es el Pokemon con mas puntos de estadistica total ('base_stat_total')? 

Para responder esta pregunta filtraremos aquellos pokemon que si tienen un hábitat asignado y encontraremos el más fuerte de cada uno

```python
# Importamos todas las librerias necesarias para mostrar y filtrar los datos correctamente
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Usamos seaborn para que le de una vista estetica de tabla a los datos que mostraremos
sns.set_theme(style="whitegrid")
%matplotlib inline

# Usando panda leemos el csv y lo guardamos en una variable, para luego mostrar que se ha cargado correctamente
df = pd.read_csv('pokemon.csv')
print(f"Dataset carregat: {df.shape[0]} files i {df.shape[1]} columnes.")
```

```python
# Ahora guardamos en una variable los pokemon que si tienen un hábitat
df_habitat = df.dropna(subset=['habitat'])
```

```python
# Para cada hábitat obtenemos la fila que indique el total de puntos de estadistica ('base_stat_total')
idx_max = df_habitat.groupby('habitat')['base_stat_total'].idxmax()
resultado = df_habitat.loc[idx_max, ['habitat', 'name', 'base_stat_total']]
```

```python
# Una vez guardamos los resultados en las variables los ordenamos para que sea ma legible
resultado = resultado.sort_values('base_stat_total', ascending=False).reset_index(drop=True)
```

```python
# Ahora mostramos el resultado
resultado
```

```python
pokemon_colors = {
    'Rayquaza':  '#3E9C35',   # Verde
    'Slaking':   '#8B6914',   # Marrón
    'Kyogre':    '#1565C0',   # Azul oscuro
    'Groudon':   '#C62828',   # Rojo
    'Tyranitar': '#4A5240',   # Gris verdoso
    'Raikou':    '#F9D000',   # Amarillo
    'Regirock':  '#A0785A',   # Marrón arenoso
    'Dragonite': '#FF9800',   # Naranja
    'Blissey':   '#F48FB1',   # Rosa
}

# Creamos la lista de colores en el mismo orden que el DataFrame resultado
colores = [pokemon_colors[name] for name in resultado['name']]

plt.figure(figsize=(10, 6))

bars = plt.barh(resultado['habitat'], resultado['base_stat_total'], color=colores)

# Añadimos el nombre del Pokémon dentro de cada barra
for bar, name in zip(bars, resultado['name']):
    plt.text(bar.get_width() - 10, bar.get_y() + bar.get_height() / 2,
             name, va='center', ha='right', fontsize=10, color='white', fontweight='bold')

# Le ponemos un titulo y tamaño a la grafica
plt.title("Pokémon més Fort per Hàbitat", fontsize=14, fontweight='bold')
plt.xlabel("Base Stat Total (Punts d'Estadística)", fontsize=11)
plt.ylabel("Hàbitat", fontsize=11)

# Finalmente lo mostramos
plt.tight_layout()
plt.show()
```
